## Batch Gradient Descent

Recall that the gradient descent algorithm is:

Repeat until convergence{

    w = w - a\*djdw

    b = b - a\*djdb

}

Where:

- $w$ is the model parameter (weight)
- \(b\) is the model parameter (bias)
- \(\alpha\) is the learning rate
- \(\frac{\partial J(w,b)}{\partial w}\) is the gradient of the cost function with respect to \(w\)
- \(\frac{\partial J(w,b)}{\partial b}\) is the gradient of the cost function with respect to b

and that:

- djdw = sum of all((w\*x[i] + b) - y[i]) \* x[i] divided by m, the number of training examples

- djdb = sum of all ((w\*x[i] + b) - y[i]) divided by m, the number of training examples

recall that:
- w*x[i] + b  = fwb(x[i])

Therefore, we can also write the gradients as:

- djdw = sum of all (fwb(x[i]) - y[i])*x[i] divided by m, the number of training examples
- djdb = sum of all (fwb(x[i]) - y[i]) divided by m, the number of training examples

## Gradient Descent aka Batch Gradient Descent

The algorithm is called batch **Batch Gradient Descent** because for each step of gradient descent, we use all the training examples. What this means is that before updating the parameters, we calculate the gradients by involving all the training examples as shown on the formula above i.e.

- djdw = sum of all (fwb(x[i]) - y[i])*x[i] divided by m, the number of training examples
- djdb = sum of all (fwb(x[i]) - y[i]) divided by m, the number of training examples

The key idea is that one gradient descent update uses the entire training set. This is what distinguishes Batch Gradient Descent from other approaches that update the parameters using only one training example for some small subset of training examples at a time.
