# 1 – Introduction to Machine Learning

## What is Machine Learning?

>A field of study that gives computers the ability to learn without being explicitly programmed. - *Arthur Samuel (1959)*

To understand ML, consider Arthur Samuel's checkers program. Rather than being explicitly programmed with the best moves to make, the program learned from playing many checkers games and gradually identified positions that increased its chances of winning.

---

## Types of Machine Learning

There are two main types of machine learning:

- Supervised Learning
- Unsupervised Learning

### Supervised Learning

>Algorithms that learn input to outuput mappings i.e. they figure out the relationships betwen inputs and outputs from past examples.

In ML call the Inputs **features** and outputs **target values**.

The goal is to train the computer to figure out how the inputs map to the outputs.

For example, suppose we have data containing the sizes of houses and the prices at which they were sold. We can train a machine learning model to learn the relationship between houses' sizes and their selling prices. Once trained, the model can then be used predict the selling price of a house given its size.

---

## Supervised Learning Problems

There are two main types of supervised learning problems.
- Regression
- Classification

### Regression

A regression problem is one in which we try to predict a continuous numerical value that can be any number out of **infinite (unlimited) number of possible values**. Using the houses example, the target value is the selling price of the house. Since the price can be one of infinitely many possible numerical values, this is therefore a regression problem.

### Classification

A classification problem is one in which we try to predict a discrete category out of a finite (limited) number of possible values. For example, predicting that an email is spam or not spam is a classification problem as the model outputs a disrete category out of two possible values (spam or not spam).

**NOTE:** In regression, the target value is a number, while in classification, the target value is a category.

For example, determining whether an email is **spam** or **not spam** is a classification problem since the target value can only be one of finite categories while predicting the price of a house given the features of a house is a regression problem since the price can be any numerical value out of infinite possible values.

Given a dataset for houses sizes in square feet and the prices at which they were sold for, we can plot the data with prices on the y axis and sizes on the x axis just to visualise the data. This helps to gain some insight into how the two variables are related.
