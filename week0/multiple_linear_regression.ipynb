## Linear Regression for multiple features

So far, our model for predicting the price of a house has used only a single feature, namely the size of the house.

In the real world, however, the price of a house depends on more than just its size. Other features that can contribute to the price of a house include:

- Number of bedrooms
- Number of bathrooms
- Age of the house
- Number of floors
- Location
- etc.

We therefore need to modify our linear regression model so that it can work with multiple features.

The version of linear regression that uses a single feature is called Univariate Linear Regression, while the version that uses multiple features is called Multiple Linear Regression.

## Multiple Linear Regression

Recall that the univariate linear regression mode:

fwb(x) = w*x[i] + b

and we trained the parameters w,b on past examples

In multiple linear regression our x is no longer a scalar ie a single value like 20, instead it will be a vector ie a list of numbers like [20, 3, 2, 5] representing say size in square meters, number of bedrooms, number of bathrooms and how old the house is in years respectively.

With multiple features say n number of features we also need n weights. That means our w is also going to change from a scalar value to vector ie a list of numbers. The important idea is that each feature has its own weight.

The models becomes:

fwb(x) = w<sub>1</sub>*x<sub>1</sub>[i] + w<sub>2</sub>*x<sub>2</sub>[i] + ... + w<sub>n</sub>*x<sub>n</sub>[i] + b



A useful distinction in the notation

There are now two indices that are important:

\(i\) identifies the training example
\(j\) identifies the feature

This distinction becomes especially important when we implement multiple linear regression because our data will now be represented as a matrix rather than a one-dimensional array.

And this is where the vectorized form becomes particularly useful:

$$ \boxed{f_{w,b}(x^{(i)}) = w\cdot x^{(i)}+b} $$

where \(w\cdot x^{(i)}\) is the dot product between the weight vector and the feature vector.
