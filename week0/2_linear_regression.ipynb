## Linear Regression Model

>A model that maps features to target values by fitting a straight line through the data. 

The features are the predictor values and the target values are the estimated/predicted values. 

Let's take the price-of-a-house prediction problem as our example. Given the sizes of houses and the prices at which they were sold for. Note that the size of the house is the feature and the price is the target value.

We can plot the data points as price vs house and try to fit a straight line through the points. We call the data a training set since that's the data we'll use to train our linear regression model on. 

The model will learn the relationship between the feature (size) and the target values (prices). We can then use the model to predict the price of a new house given its feature i.e. size.

Notation:
We will use x to represent feature variables  and y to represent the target variables. m is going to be used to denote the number of examples in the training set i.e. the size of the training set.

| x  |  y  |
| -- |  -- |
| 20 | $100 |
| 25 | $150 |
| 27 | $156 |
| 21 | $ 97 |

In this case m = 4

In order to identify the training examples we use the superscript i eg x<sup>i</sup> and y<sup>i</sup> to say the i<sup>th</sup> example. 

For example; 

The first example is (x<sup>1</sup>,y<sup>1</sup>) = (20,100)

The second example is (x<sup>2</sup>,y<sup>2</sup>) = (25,150)

In ML, the convention is that y-hat is the predicted value. The letter y is the true value in the training set. 
So given x, the model will output y-hat.

f<sub>w,b</sub>(x) = $wx + b$ 

Thus, 

y-hat = f<sub>w,b</sub>($x$)


$w$ and $b$ are called the paramaters of the model. In ML, paramaters aka weights or coefficients are variables you can adjust during training in order to improve the model.

Notice that $w$ is the slope of the line and $b$ is our y-intercept.

**Cost function formula**

Recal that;

f<sub>w,b</sub>($x$<sup>i</sup>) = $wx$<sup>i</sup> + $b$

Now the question is how do we find the values of $w$ and $b$ so the models y-hat<sup>i</sup> that is the predicted value is very close to the true value that is y<sup>i</sup>?

The cost function calculates how far the model's predictions are from the true values.

Thus,

1/2m $\sum$ <sub>i=1 to m</sub>(y-hat<sup>(i)</sup> - y<sup>(i)</sup>)<sup>2</sup>

This is known as Squared Error cost function.

Why do we square the error?

The prediction error for a training example is:

(y-hat<sup>(i)</sup> - y<sup>(i)</sup>)

Some errors will be positive and others negative. If we simply added the errors together, they could cancel each other out.
For example:

[
(+5) + (-5) = 0
]

This would incorrectly suggest that there was no error.

By squaring the errors:

[
(+5)^2 = 25
]

[
(-5)^2 = 25
]

Why divide by (2m)?

Dividing by (m) gives us the mean squared error across all training examples.

The additional (2) is mainly included because it makes the mathematics of gradient descent simpler when we take derivatives. It does not change which values of (w) and (b) minimize the cost.

We'll denote the cost function as J(w,b).

Thus,

J(w,b) = 1/2m $\sum$ <sub>i=1 to m</sub>(y-hat<sup>(i)</sup> - y<sup>(i)</sup>)<sup>2</sup>

Substituting y-hat<sup>i</sup> with wx<sup>(i)</sup>) + b we get our cost function as:

J(w,b) = 1/2m $\sum$ <sub>i=1 to m</sub>(wx<sup>(i)</sup>+b-y<sup>(i)</sup>)

Note that the squared error cost function is just one of several cost functions available.


## Cost function Intuition.

Recall that we need the cost function to determine how well our choice of the values of the parameters (w,b) fits the trainining data during training. In other words, since the cost function measures how far the y-hat i.e. the models predicted values are during each iteration of training are to y i.e. the true values in the training set, we can keep on adjusting the paramaters in the next interations to try to get our models predictions closer and closer to the true values.

Therefore, in the training process, we try to minimize J(w,b). To get some intuition of the const function, we can plot the cost vs parameters graph just to see how the cost changes with respect to the parameters. Now to simplify the example, let's set $b$ equal to $0$. On the x axis we can have different values of w and then compute the cost.

Example:

Let's say we take our choice of w to be 4.5, we would then calculate the cost as J(w) = 1/2m Sum((fw(x<sup>i</sup>) - y<sup>i</sup>)<sup>2</sup>)

Recall that; fw(x<sup>i</sup>) = wx<sup>i</sup> 

So J(w) = 1/2m Sum((wx<sup>i</sup>) - y<sup>i</sup>)<sup>2</sup>)

Let's take our training set from before:
| x  |  y  |
| -- |  -- |
| 20 | $100 |
| 25 | $150 |
| 27 | $156 |
| 21 | $ 97 |

J(4.5) = 1/(2*4) * (((4.5*20) - 100)<sup>2</sup> + ((4.5*25)-150)<sup>2</sup> + ((4.5*27)-156)<sup>2</sup>) + ((4.5*21)-97)<sup>2</sup>) 

  = 1/8 * ((90-100)<sup>2</sup> + (112.5-150)<sup>2</sup> + (121.5-156)<sup>2</sup> + (94.5-97)<sup>2</sup>)

  = 0.125 * ((-10)<sup>2</sup> + (-37.5)<sup>2</sup> + (-34.5)<sup>2</sup> + (-2.5)<sup>2</sup>)

  = 0.125 * (100 + 1406.25 + 1190.5 + 6.25)

  = 0.125 * 2703

  = 337.875

  We can repeat then adjust our choice of w and re-calculate the the cost.

  Let's now take the value of 3.5 as our choice of w;

  J(3.5) = 1/(2*4) * (((3.5*20) - 100)<sup>2</sup> + ((3.5*25)-150)<sup>2</sup> + ((3.5*27)-156)<sup>2</sup>) + ((3.5*21)-97)<sup>2</sup>) 
  
  = 1142.59375

  This tells us that, for these particular values of \(w\), \(w=4.5\) produces predictions that are closer to the actual values than \(w=3.5\).

We could continue trying different values of \(w\), calculating the cost each time, and looking for the value of \(w\) that produces the smallest possible cost.

This is the basic idea behind minimizing the cost function. 

How can we systematically find the value of w or in the general case w and b that minimizes the cost function J(w)? **Gradient Descent algorithm** allows us to achieve exactly this.
